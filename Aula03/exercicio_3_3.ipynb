```python
from secrets import choice

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dados = pd.read_csv('vendas_shampoo.txt')
dados.head()
```

```python
plt.figure(figsize=(8,2))
plt.title("Vendas ao longo dos meses")
sns.lineplot(x='Month', y='Sales', data=dados)
plt.xticks([ dados['Month'].values[int(len(dados['Month'].values) / 10 * i)] for i in range(0, 10)])
plt.show()
```

As vendas tendem a crescer ao longo dos meses.
