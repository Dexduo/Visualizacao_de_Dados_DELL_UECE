#### Gráficos de Dispersão para mais de duas variáveis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style('darkgrid')
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
plt.figure(figsize=(8, 3))
plt.title("Relação entre valor da conta e gorgeta por gênero")
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips)
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.title("Relação entre valor da conta e gorgeta por dia da semana")
sns.scatterplot(x='total_bill', y='tip', hue='day', data=tips)
plt.show()
```

```python
sns.relplot(x='total_bill', y='tip', col='day', data=tips)
plt.show()
```

```python
sns.lmplot(x='total_bill', y='tip', hue='sex', data=tips)
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.title("Relação entre valor da conta e gorgeta pelo tamanho da mesa")
sns.scatterplot(x='total_bill', y='tip', hue='size', data=tips)
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.title("Relação entre valor da conta, gorgeta e tamanho da mesa")
sns.scatterplot(x='total_bill', y='tip', size='size', sizes=(15,200), data=tips)
plt.show()
```

```python
plt.figure(figsize=(16,9))
plt.title("Relação entre valor da conta, gorgeta, tamanho da mesa, fumante e sexo")
sns.scatterplot(x='total_bill', y='tip', size='size', sizes=(15, 200), hue='sex', style='smoker', data=tips)
plt.show()
```
