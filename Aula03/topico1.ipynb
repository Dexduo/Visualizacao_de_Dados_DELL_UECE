```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
casas_alugar = pd.read_csv("dados.txt")
casas_alugar.head()
```

```python
casas_alugar['area'].describe()
```

```python
plt.figure(figsize=(10, 5))
sns.set_style('darkgrid')
plt.title("Distruição da área em m2 das casas")
plt.hist(casas_alugar['area'], bins=10)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.set_style('darkgrid')
plt.title("Distruição da área em m2 das casas")
plt.hist(casas_alugar['area'], bins=25)
plt.show()
```
