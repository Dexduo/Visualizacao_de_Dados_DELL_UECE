```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style('darkgrid')
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
plt.figure(figsize=(8, 4))
plt.title("Relação entre o valor da conta e a gorgeta")
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.title("Relação entre o valor da conta e a gorgeta com linha de regressão")
sns.regplot(x='total_bill', y='tip', data=tips)
plt.show()
```
