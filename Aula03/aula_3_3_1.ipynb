### Trabalhando com dados temporais

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style("darkgrid")
```

```python
dados = pd.read_csv("dados_temporais.txt")
dados.head()
```

#### Variação da temperatura ao longo do tempo

```python
plt.figure(figsize=(10,5))
plt.title("Variação da temperatura ao longo do tempo")
sns.scatterplot(x='date', y='Temperature', data=dados )

plt.xticks(0)
```

#### Gráfico de linha

```python
dados['date'] = pd.to_datetime(dados['date'])
plt.figure(figsize=(10,5))
plt.title("Gráfico de linha da temperatura ao longo do tempo")
sns.lineplot(x='date', y='Temperature', data=dados)
plt.show()
```

#### Dados de incidência do HIV ao longo dos anos nos países

```python
dados_hiv = pd.read_csv('hiv_data.txt')
dados_hiv.head()
```

```python
plt.figure(figsize=(10,5))
plt.title("Variação da taxa média de HIV global")
sns.lineplot(x='date', y='HIV Rate', data=dados_hiv)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.title("Variação de taxa média de HIV global por nível de renda")
sns.lineplot(x='date', y='HIV Rate', hue='Income Level', errorbar=None, data=dados_hiv)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.title("Variação da taxa média de HIV global por países")
sns.lineplot(x='date', y='HIV Rate', hue='country', errorbar=None, legend=False, data=dados_hiv)
plt.show()
```
