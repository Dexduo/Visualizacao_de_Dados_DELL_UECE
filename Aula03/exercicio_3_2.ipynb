```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

#### Importação do dataset MPG

```python
carros = sns.load_dataset('mpg')
carros.head()
```

#### Relação entre aceleração e consumo de combustível por país de origem

```python
plt.figure(figsize=(6,6))
plt.title("Relação entre aceleração e consumo de combustível por país de origem")
sns.scatterplot(x='acceleration', y='mpg', style='origin',hue='origin', data=carros)
plt.show()
```

```python
# plt.title("Relação entre aceleração e consumo de combustível por país de origem")
sns.relplot(x='acceleration', y='mpg', col='origin', data=carros)
```

A correlação existe e é baixa positiva.</br>
Difere um pouco a depender do país de origem.

#### Relação entre aceleração e consumo de combustível por ano do modelo

```python
plt.figure(figsize=(8,8))
plt.title("Relação entre aceleração e consumo de combustível por ano do modelo")
sns.scatterplot(x='acceleration', y='mpg', hue='model_year', data=carros)
plt.show()
```

É possível observar que ao passar dos anos os carros ganharam potência e junto a isso aumentaram o consumo.
