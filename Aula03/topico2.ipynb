```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('dados.txt')
df.head()
```

```python
plt.figure(figsize=(10,5))
plt.title("Distribuição área de imóveis")
sns.distplot(df['area'])
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.title("Distribuição de área de imóveis")
sns.kdeplot(df['area'], shade=True)
plt.show()
```
