```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

#### Carregamento dos dados

```python
df = pd.read_csv('dados.txt')
df.head()
```

#### Distribuiçao de Valor de Aluguel

```python
plt.figure(figsize=(10, 5))
plt.title("Distrbuição de Valor de Aluguel")
sns.distplot(df['rent amount (R$)'])
plt.show()
```
