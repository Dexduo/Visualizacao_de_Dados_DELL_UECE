```python
import pandas as pd
```

```python
df = pd.read_csv('drinks.txt')
```

```python
df
```

#### Média e Desvio Padrão do consumo de cerveja, licor, vinho e total de álcool

```python
media_consumo_cerveja = df['beer_servings'].mean()
dp_consumo_cerveja = df['beer_servings'].std()
print('Média consumo de litros de cerveja:',media_consumo_cerveja)
print('Desvio Padrão consumo de litros de cerveja:',dp_consumo_cerveja)
```

```python
media_consumo_licor = df['spirit_servings'].mean()
dp_consumo_licor = df['spirit_servings'].std()
print('Média consumo de litros de licor:',media_consumo_licor)
print('Desvio Padrão consumo de litros de licor:',dp_consumo_licor)
```

```python
media_consumo_vinho = df['wine_servings'].mean()
dp_consumo_vinho = df['wine_servings'].std()
print('Média consumo de litros de vinho:',media_consumo_vinho)
print('Desvio Padrão consumo de litros de vinho:',dp_consumo_vinho)
```

```python
media_consumo_alcool = df['total_litres_of_pure_alcohol'].mean()
dp_consumo_alcool = df['total_litres_of_pure_alcohol'].std()
print('Média consumo de litros de alcool:',media_consumo_alcool)
print('Desvio Padrão consumo de litros de alcool:',dp_consumo_alcool)
```

##### Outra maneira

```python
md_cons_cerv = df.describe()['beer_servings']['mean']
dp_cons_cerv = df.describe()['beer_servings']['std']
print('Média de consumo de cerveja:', md_cons_cerv)
print('Desvio Padrão de consumo de cerveja:', dp_cons_cerv)
```

#### Qual o consumo de licor na França?

```python
cons_licor_franca =  df[df['country'] == 'France']['spirit_servings'].values[0]
print('O consumo de licor na França é:', cons_licor_franca)
```

#### Quais os 5 países nos quais se consome mais vinho

```python
top5_paises_vinho =  df.sort_values(by='wine_servings', ascending=False).head(5)['country'].values
print('Os 5 países que mais se consome vinho são: ', str(', ').join(top5_paises_vinho))
```

#### Quais os países nos quais o consumo de cerveja está acima da média?

```python
media_cerveja = df['beer_servings'].mean()
paises_media_cerveja = df[df['beer_servings'] > media_cerveja]['country'].values
paises_media_cerveja = str(', ').join(paises_media_cerveja)
print('Os países em que o consumo de cerveja está acima da média são:', paises_media_cerveja)
```

#### Países com consumo de álcool maior que o Brasil, ordem crescente

```python
consumo_alcool_brasil = df[df['country'] == 'Brazil']['total_litres_of_pure_alcohol'].values[0]
consumo_alcool_brasil
```

```python
mais_que_brasil = df[df['total_litres_of_pure_alcohol'] > consumo_alcool_brasil]
mais_que_brasil = mais_que_brasil.sort_values(by='total_litres_of_pure_alcohol')
mais_que_brasil
```
