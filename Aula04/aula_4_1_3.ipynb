### Gráfico de Violino

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
plt.figure(figsize=(10, 5))
plt.title("Distribuição de idade por classes com Gráfico de Violino")
sns.violinplot(x='class', y='age', hue='class', data=titanic)
plt.show()
```

```python
titanic3 = titanic[titanic['class'] =='Third']
sns.kdeplot(x='age', data=titanic3)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.title("Distribuição de idade por classes com Gráfico de Violino")
sns.violinplot(x='class', y='age', hue='class', data=titanic)
sns.swarmplot(x='class', y='age', color='black', size=3, data=titanic)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.title("Distribuição de idade por classes com Gráfico de Violino")
sns.violinplot(x='class', y='age', hue='sex', split=True, data=titanic)
plt.show()
```

```python
sns.catplot(x='class', y='age', hue='sex', col='survived', kind='violin', split=True, data=titanic)
plt.show()
```
