```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL.ImImagePlugin import split
%matplotlib inline
```

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic['embark_town'].unique()
```

```python
plt.title("Distribuição de valor de passagem de acordo com a cidade de embarque", y=1.05)
# sns.catplot(x='embark_town', y='fare', hue='sex', col='survived', split=True, kind='violin', data=titanic)
sns.violinplot(x='embark_town', y='fare', data=titanic)
plt.show()
```

```python
sns.catplot(x='embark_town', y='fare', col='survived', kind='violin', data=titanic).fig.suptitle("Distribuição de valor de passagem de acordo com a cidade de embarque", y=1.05)
plt.show()
```
