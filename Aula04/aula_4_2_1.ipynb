#### Distribuições Bivariavadas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style("darkgrid")
```

```python
iris = sns.load_dataset("iris")
iris.head()
```

```python
sns.kdeplot(x='sepal_length', data=iris)
plt.show()
```

```python
sns.jointplot(x='sepal_length', y='petal_length', kind='reg', data=iris)
```

```python
sns.jointplot(x='sepal_length', y='petal_length', kind='hex', color='black', data=iris)
```

```python
sns.jointplot(x='sepal_length', y='petal_length', kind='kde', fill=True, data=iris)
```
