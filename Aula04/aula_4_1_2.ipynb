### Gráfico de Caixa

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style("darkgrid")
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic['age'].describe()
```

```python
plt.figure(figsize=(10, 5))
plt.title("Distribuição de idades de acordo com a classe com Gráfico de Caixa")
sns.boxplot(x='class', y='age', hue='class', data=titanic)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.title("Distribuição de idades de acordo com a classe com Gráfico de Caixa")
sns.boxplot(x='class', y='age', hue='sex', data=titanic)
plt.show()
```

```python
sns.boxenplot(x='class', y='age', hue='sex', data=titanic)
plt.show()
```

```python
sns.catplot(x='class', y='age', hue='sex', col='survived', kind='boxen', data=titanic)
plt.show()
```
