### Gráfico de Dispersão Gráfico Categórico

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style("darkgrid")
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

Vamos focar na idade (age), classe da passagem (class), sexo (sex), verificar se o passageiro sobreviveu (survived).

```python
plt.figure(figsize=(10,5))
plt.title("Relação entre idade e classe da passagem")
sns.stripplot(x='class', y='age', hue='class', data=titanic)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.title("Relação entre idade e classe da passagem")
sns.swarmplot(x='class', y='age', hue='sex', data=titanic)
plt.show()
```

```python
sns.catplot(x='class', y='age', hue='sex', col='survived', kind='swarm', data=titanic)
plt.show()
```
