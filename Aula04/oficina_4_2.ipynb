```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
mpg = sns.load_dataset('mpg')
mpg.head()
```

```python
sns.pairplot(mpg, hue='origin')
```

Observando de maneira empírica, as váriaveis que tem maior correlação entre si são mpg, horsepower, weight.

Foi possível distinguir o país de origem pois os dados ainda são visualmente diferenciáveis.

```python
sns.pairplot(mpg, hue='origin', kind='reg', vars=['mpg', 'weight', 'horsepower'])
```
