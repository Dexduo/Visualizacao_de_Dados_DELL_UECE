```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
especie = ['Cachorro', 'Gato', 'Pássaro']
quantidade = [110, 90, 30]
```

```python
plt.title("Proporção de espécies")
plt.pie(x= quantidade, labels=especie)
plt.show()
```

```python
plt.figure(figsize=(8,8))
# plt.style.use('seaborn-pastel')
plt.title('Proporção de espécies')
plt.pie(x=quantidade, labels=especie, autopct='%1.2f%%')
plt.show()
```

```python
races = ['Poodle', 'Golden Retriever', 'Bulldog', 'Pastor Alemão', 'Labrador', 'Sem Raça Definida']
quant_races = [25, 15, 5, 20, 25, 10]
```

```python
plt.title('Proporção de raças cães')
plt.pie(x=quant_races, labels=races, autopct='%1.2f%%')
plt.show()
```

```python
plt.pie(x=quant_races, labels=races, autopct='%1.2f%%', startangle=90, pctdistance=0.85)
my_circle = plt.Circle((0,0), 0.7, color='white')
p = plt.gcf()
p.gca().add_artist(my_circle)
plt.title("Proporção de raças de cães")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
games = pd.read_csv('vgsales.csv')
games.head()
```

```python
genero = games.drop(['Name', 'Rank', 'Platform', 'Year', 'Publisher'], axis=1).groupby('Genre').sum()
#obs.: deletei temporariamente as colunas de dados categóricos
genero
```

```python
plt.pie(x=genero['Global_Sales'], labels=genero.index, autopct='%1.1f%%', startangle=90, pctdistance=0.85)
plt.show()
```

```python
plt.pie(x=genero['Global_Sales'], labels=genero.index, autopct='%1.1f%%', startangle=90, pctdistance=0.85)
my_circle = plt.Circle((0,0), 0.7, color='white')
p = plt.gcf()
p.gca().add_artist(my_circle)
plt.show()
```

```python
genero = genero.sort_values(by='Global_Sales', ascending=False)
genero
```

```python
porcentagens = genero['Global_Sales']/genero['Global_Sales'].sum()*100
```

```python
# plt.style.use('seaborn-whitegrid')
sns.set_style('whitegrid')
plt.figure(figsize=(10,6))
plt.bar(genero.index, porcentagens)
plt.title("Porcentagem de Vendas Globais por Gênero")
plt.show()
```
