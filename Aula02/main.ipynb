```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = {
    'categorias': ['A', 'B', 'C', 'D'],
    'valores': [10, 20, 15, 5]
}

df = pd.DataFrame(dataset)
categorias = df['categorias'].values
valores = df['valores'].values
```

```python
plt.bar(categorias, valores)
plt.xlabel('Grupos')
plt.ylabel('Tamanho')
plt.title('Gráfico de Barras Categorias X Valores')
plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.bar(categorias, valores, color='darkred')
plt.xlabel('Grupos')
plt.ylabel('Tamanho')
plt.title('Gráfico de Barras Categorias x Valores')
plt.show()
```

#### Capturando dados do json e exibindo com Matplotlib

```python
url_json = 'https://raw.githubusercontent.com/emanueles/datavis-course/master/assets/files/observable/movies.json'
filmes = pd.read_json(url_json)
filmes.head(5)
```

```python
filmes.describe()
```

```python
contagem = filmes.groupby(by='Genre').count()
contagem
```

```python
indices = contagem.index.values
quantidade = contagem['Film'].values
```

```python
plt.figure(figsize=(6, 3))
plt.bar(indices, quantidade)
plt.show()
```

#### Utilzando a Seaborn

```python
sns.set(style='whitegrid')
sns.countplot(x='Genre', data=filmes)
plt.show()
```

```python
sns.set(style='darkgrid')
plt.figure(figsize=(6, 3))
plt.title('Quantidade de Filmes por Gênero')
sns.countplot(x='Genre', data=filmes)
plt.show()
```

```python
plt.title('Quantidade de Filmes por Gênero')
sns.countplot(y='Genre', color='r', data=filmes)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.title('Bilheteria média por gênero')
sns.barplot(x='Genre', y='Worldwide_Gross_M', color='b', ci='sd', data=filmes)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.title('Bilheterial Total por Gênero')
sns.barplot(x='Genre', y='Worldwide_Gross_M', color='orange', estimator=sum, ci=None, data=filmes)
plt.show()
```
